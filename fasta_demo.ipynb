# ```fasta.py``` Functions Demonstration

This JupyterNotebook demonstrates how the functions found in ```fasta.py``` work. 

````python
from fasta import *
````

## Downloading FASTA data

```get_seqs_from_fasta``` reads data from a FASTA file and constructs a dictionary from it.

Each entry contains the sequence ID, its chromosome, its start and stop positions, and the sequence.

````python
seqs_dico = get_seqs_from_fasta("Sequence_by_Peaks_1.fasta")

# print info on "seqs_dico", resulting data from get_seqs_from_fasta()
print("information on `seqs_dico`")
print(f"\ttype: {type(seqs_dico)}")
print(f"\tsize: {len(seqs_dico)}\n")


# print example sequence entry
print("ex. first entry")
fst = list(seqs_dico.items())[0]
print(f"\t{fst[0]}")
for x in fst[1].items() :
    if x[0] != "Chromosome" : print(f"\t\t{x[0]}\t\t{x[1]}")
    else : print(f"\t\t{x[0]}\t{x[1]}")
````

## Pre-Processing FASTA data

### Python Lists and Dictionaries

```id_seq()``` reduces the full dictionary to just the sequence id and the sequence.

````python
ids_seqs = id_seq(seqs_dico)
print("information on `ids_seqs`")
print(f"\ttype: {type(ids_seqs)}")
print(f"\tsize: {len(ids_seqs)}\n")

print("ex. first entry")
print(f"\t{list(ids_seqs.items())[0]}")
````

```seq_only``` creates a sequence-only list.

````python
seqs = seq_only(seqs_dico)
print("information on `seqs`")
print(f"\ttype: {type(seqs)}")
print(f"\tsize: {len(seqs)}\n")

print("ex. first entry")
print(f"\t{seqs[0]}")
````

### Pandas Dataframe

```seqs_info_to_pd()``` converts the Python dictionary into a Pandas Dataframe.

````python
seqs_pd = seqs_info_to_pd(seqs_dico)
display(seqs_pd.head())

print(f"\ninformation on `seqs_pd`")
print(f"\ttype: {type(seqs_pd)}")
print(f"\tsize: {seqs_pd.shape}")
print(f"\tcolumns: {list(seqs_pd.columns)}")
````

````python
ids_seqs_pd = seqs_pd.drop(["chromosome","start","stop"], axis=1)
display(ids_seqs_pd.head())
````

````python
seqs_only_pd = seqs_pd["sequence"]
display(seqs_only_pd.head())
````
