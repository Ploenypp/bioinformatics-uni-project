## Preparation & Pre-Processing

### FASTA Data

This FASTA data was obtained through ChIP-Seq, ensuring that these sequences contained the same (or a similar) DNA-protein binding site, in this case, a transcription factor binding site.  However, what motif (or sub-sequence) that makes up the site is still to be identified.

```python
seqs_dico = get_seqs_from_fasta("Sequence_by_Peaks_1.fasta")
seqs_pd = seqs_info_to_pd(seqs_dico)
display(seqs_pd.head())
```

```python
seqs = list(seqs_pd["sequence"])
```

### Candidate Kmers Generation

Instead of blindly and naively searching through all of the given sequences, first generating candidate motifs and eliminating the ones that do not satisfy the complexity or TA nucleotide propotion criteria can reduce algorithmic complexity.

```python
kmers = candidate_kmers(7, 3, 2, 1)
```

## Search Algorithms

The next step is to identify which of the theorectically biologically significant candidate motifs are most frequently present within the given sequences. This can be done with a number of different algorithms.

### HashTable

```python
ht = hashtable(seqs, kmers)
```

```python
df_ht = pd.DataFrame({"kmers":list(ht.keys()),"freqs":list(ht.values())})
display(df_ht.head(5))
```

_Note how the 2nd most frequent kmer is reverse complementary to the 1st most frequent kmer._

### Median String

```python
ms = medianString(seqs, kmers)
```

```python
df_ms = pd.DataFrame({"kmers":list(ms.keys()),"distances":list(ms.values())})
display(df_ms.head(5))
```
